```python
import pandas as pd

products = pd.read_csv("../data/output/product_details.csv")

print(len(products))
```

```python
products = products.drop_duplicates(subset=["URL"])
print(len(products))

```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))
plt.hist(products["Rating"], bins=10)
plt.title("Distribution of product ratings")
plt.xlabel("Rating")
plt.ylabel("Number of products")
plt.show()

plt.figure(figsize=(6, 4))
plt.hist(products["Price"].dropna(), bins=100)
plt.title("Distribution of product prices")
plt.xlabel("Price (USD)")
plt.ylabel("Number of products")
plt.show()


```
